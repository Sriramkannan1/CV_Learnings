```python
img4=cv2.rotate(img,cv2.ROTATE_90_COUNTERCLOCKWISE)
cv2_imshow(img4)
```

```python
img5=cv2.rotate(img,cv2.ROTATE_180)
cv2_imshow(img5)
```

```python
img6=cv2.flip(img,0)
cv2_imshow(img6)
```

```python
img7=cv2.flip(img,1)
cv2_imshow(img7)
```
